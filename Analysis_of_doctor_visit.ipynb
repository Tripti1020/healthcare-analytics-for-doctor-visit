```python
#importing important libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Load the dataset and Display first 15 rows

```python
df=pd.read_csv("DoctorVisits - DA.csv")
df.head(15)
```

List the number of rows and columns

```python
print("The number of rows in the dataset: ",df.shape[0])
print("The number of columns in the dataset: ",df.shape[1])
```

Display the names of columns with their datatypes 

```python
df.info()
```

Find out the total number of people based on their count of illness

```python
df["illness"].value_counts()
```

```python
df["gender"].value_counts()
```

Visualize and analyse the maximum, minimum and median of income

```python
y=list(df.income)
plt.boxplot(y)
plt.show
```

Find the number of days and reduced activity of male and female separately due to illness

```python
df.groupby(["gender","reduced"]).mean(numeric_only=True)
```

Find out the correlation  between different variables in the given dataset

```python
plt.figure(figsize=(10,10))
sns.heatmap(df.corr(numeric_only=True),cbar=True,annot=True,cmap="Blues",)
plt.title("Correlation Heatmap")
```

Analyse how the income of a patient affects the no of visits to the hospital

```python
plt.figure(figsize=(10,10))
plt.scatter(x="income",y="visits",data=df)
plt.xlabel("Income")
plt.ylabel("visits")
plt.title("No of visits based on income")
plt.show()
```

Count and Visualize the number of males and females affected by illness

```python
sns.histplot(df.gender,bins=2)
```

Visualize the percentage of people  getting govt health insurance due to low income, old age and also percentage of people getting private health insurance 

```python
#% of people getting govt health insurance due to low income
label=["yes", "no"]
Y=df[df["freepoor"]=="yes"]
N=df[df["freepoor"]=="no"]
x=[Y.shape[0],N.shape[0]]
plt.figure(figsize=(5,5))
plt.pie(x,labels=label)
plt.title("% of people getting govt health insurance due to low income")
plt.show()
#% of people getting private health insurance due to low income
Y=df[df["private"]=="yes"]
N=df[df["private"]=="no"]
x=[Y.shape[0],N.shape[0]]
plt.figure(figsize=(5,5))
plt.pie(x,labels=label)
plt.title("% of people getting private health insurance due to low income")
plt.show()
#% of people getting govt health insurance due to old age,disability or veteran status
Y=df[df["freerepat"]=="yes"]
N=df[df["freerepat"]=="no"]
x=[Y.shape[0],N.shape[0]]
plt.figure(figsize=(5,5))
plt.pie(x,labels=label)
plt.title("% of people getting govt health insurance due to old age,disability or veteran status")
plt.show()
```

Plot a horizontal bar chart to analyze the reduced days of activity due to illness based on Gender

```python
db=df.groupby("gender")["reduced"].sum().to_frame().reset_index()
plt.barh(db["gender"],db["reduced"],color=["cornflowerblue","lightseagreen"])
plt.title("Bar Chart")
plt.xlabel("Gender")
plt.ylabel("Reduced activity")
plt.show()
```
